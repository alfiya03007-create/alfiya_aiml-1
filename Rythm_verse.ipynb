```python
import numpy as np
import pandas as pd

# 1. Load the data
df = pd.read_csv("Document from Alfiya.csv")


# 2. Helper function to let the user choose from a list
def choose(title, options):
    print("\n" + title)
    for i, option in enumerate(options, start=1):
        print(f"  {i}. {option}")
    while True:
        choice = input("Enter number: ")
        if choice.isdigit() and 1 <= int(choice) <= len(options):
            return options[int(choice) - 1]
        print("Invalid choice, try again.")


print("=== Music Recommendation System ===")

# 3. Take user input
mood = choose("Select your mood:", sorted(df["mood"].unique()))
genre = choose("Select your genre:", sorted(df["genre"].unique()))
language = choose("Select your language:", sorted(df["language"].unique()))

# 4. Filter songs using pandas
result = df[(df["mood"] == mood) & (df["genre"] == genre) & (df["language"] == language)]

# If no exact match, relax the genre
if result.empty:
    print("\nNo exact match found. Showing songs of your mood and language.")
    result = df[(df["mood"] == mood) & (df["language"] == language)]

# 5. Show up to 3 random songs using numpy
if result.empty:
    print("Sorry, no songs found.")
else:
    count = min(3, len(result))
    picks = np.random.choice(len(result), size=count, replace=False)

    print("\nRecommended songs for you:\n")
    for i in picks:
        row = result.iloc[i]
        print(f"{row['song']} - {row['artist']}  ({row['genre']}, {row['language']}, {row['mood']})")
```
